```python
#Import libraries

import pandas as pd 
import seaborn as sns 
import numpy as np


import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import matplotlib
plt.style.use('ggplot')
from matplotlib.pyplot import figure 

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8) # Adjusts the configuration of the plots that I Will create 

pd.options.mode.chained_assignment = None

# Read in the data 
df = pd.read_csv(r'C:\Descargas\movies\movies.csv', encoding='latin-1')
```

```python
# Look the data 

df.head()
```

```python
# if there is any missing data 

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col, round(pct_missing)))
```

```python
# Data types for columns

df.dtypes
```

```python
# Change data type for columns 

df['budget'] = df['budget'].astype('int64')

df['gross'] = df['gross'].astype('int64')
```

```python
#Create correct Year column 

df['yearcorrect'] = df['released'].astype(str).str[:4]

df.head()
```

```python
df = df.sort_values(by=['gross'], inplace=False, ascending=False)
```

```python
pd.set_option('display.max_rows', None)
```

```python
# Scatter plot with budget vs gross

plt.scatter(x=df['budget'],y=df['gross'])

plt.title('Budget vs Gross Earnings')

plt.xlabel('Gross Earnings')

plt.ylabel('Budget for Film')

plt.show()
```

```python
df_numerized = df.copy()


for col_name in df_numerized.columns:
    if not pd.api.types.is_numeric_dtype(df_numerized[col_name]):
        df_numerized[col_name]= df_numerized[col_name].astype('category').cat.codes

        
df_numerized.head()
```

```python
correlation_matrix = df_numerized.corr(numeric_only=True)

sns.heatmap(correlation_matrix, annot=True)

plt.title('Correlation Matric for Numeric Feactures')

plt.xlabel('Movie Features')

plt.ylabel('Movie Features')

plt.show()
```

```python
df_numerized.corr()
```

```python
correlation_mat = df_numerized.corr()

corr_pairs = correlation_mat.unstack()

corr_pairs
```

```python
sorted_pairs = corr_pairs.sort_values()

sorted_pairs
```

```python
high_corr = sorted_pairs[(sorted_pairs) > 0.5]

high_corr
```
